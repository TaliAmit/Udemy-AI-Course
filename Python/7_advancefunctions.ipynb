```python
# Define a recursive function to calculate the nth Fibonacci number using memoization.
# Test the function with different inputs.

# Fibonacci: [1,1,2,3,5,8,13...]
# Each number is the sum of the two preceding numbers: 1+1 = 2, 1+2 = 3 etc.

# gets only positive numbers
def fib(num):
    if num == 1:
        return 1
    if num == 2:
        return 1
    else:
        return fib(num - 1) + fib(num - 2)

def calculate_fibonacci_nth_element(n):
    if n >= 1:
        print(f"Fibonacci[{n}]= {fib(n)}")
    else:
        print("Error: n is not a positive number")
        
nums = [-1,4,2,6,1]
for num in nums:
    calculate_fibonacci_nth_element(num)
```

```python
# Define a function that takes two arguments, a and b,
# where b is a dictionary with a default value of an empty dictionary.
# The function should add a new key-value pair to the dictionary and return it.
# Test the function with different inputs.

# a is a pair {key: val} and b is the dictionary
def add_to_dictionary(a,b={}):
    key, val = a
    if b.get(key) is None:
        b[key] = val
    return b

dictionary1 = {'1':10, '2':20, '3':30}
elem1 = ('1', 40)
elem2 = ('4', 40)
elem3 = (None,None)

print(add_to_dictionary(elem1))
print(add_to_dictionary(elem1, dictionary1))
print(add_to_dictionary(elem2))
print(add_to_dictionary(elem2, dictionary1))
print(add_to_dictionary(elem3))
print(add_to_dictionary(elem3, dictionary1))   
```

```python
# Define a function that takes a variable number of keyword arguments
# and returns a dictionary containing only those key-value pairs where the value is an integer.
# Test the function with different inputs.

def filter_integer_values(**kwargs):
    return {key: value for key, value in kwargs.items() if isinstance(value, int)}     


def is_val_an_int(**kwargs):
    result = {}
    for key, val in kwargs.items():
        if isinstance(val, int):
                result[key] = val
    return result

print(filter_integer_values(a=1,b="r", c="5",str1=4))    
print(is_val_an_int(a=1,b="r", c="5",str1=4)) 
```

```python
# Define a function that takes a list of integers and returns their average.
# The function should handle any errors that occur (e.g., empty list) and return None in such cases.
# Test with different inputs.


# If someone passes a number instead of a list      - len() fails.
# If someone passes a list with text [1, "hello"]   - sum() fails.

def calc_avg(lst):
    try:
        lst = [elem for elem in lst if not isinstance(elem, bool)] # Handles a bool list (True = 1, False = 0)
        return sum(lst) / len(lst)
        
    except (TypeError, ZeroDivisionError):
        return None
        
    # TypeError - lst not iterable (e.g., a number) or contains unsummable types (e.g., strings).
    # ZeroDivisionError - Raised if the list is empty (or becomes empty after filtering), causing a division by zero.
    
    # instead of:
    
def calc_avg2(lst):
    try:
        # Check if the list is empty (or None)
        if not lst:
            return None
        
        # Filter for integers (and exclude booleans, which Python treats as ints)
        nums = [elem for elem in lst if isinstance(elem, int) and not isinstance(elem, bool)]
        
        # If the list had elements but no valid integers, return None to avoid division by zero
        if not nums:
            return None
            
        # Calculate and return the average
        return sum(nums) / len(nums)
        
    except TypeError:
        # Handles cases where the input is not iterable at all (e.g., passing a single number)
        return None
```

```python
# Define a function that generates an infinite sequence of Fibonacci numbers.
# Test by printing the first 10 numbers in the sequence.

# This is better:
def fib_gen():
    num1, num2 = 1, 1
    
    while True:
        yield num1
        num1, num2 = num2, num1 + num2 # execution order: right to left

limit = 10    
gen = fib_gen()

for _ in range(limit):
    print(next(gen))

# another solution:

def fib(n):
    if n < 1:
        return 0
    if n == 1 or n == 2:
        return 1
    else:
        return fib(n-1) + fib(n-2)

def infinite_seq_of_fibonacci():
    n = 1
    while True:
        print(fib(n))
        n += 1
        
# infinite_seq_of_fibonacci()        
                 
```

```python
# Define a function that takes a list of mixed data types (integers, strings, and floats)
# and returns three lists: one containing all the integers,
# one containing all the strings, and one containing all the floats.
# Test with different inputs.

def mixed_lists_separation(mixed_lst):
    floats, integers, strings = [], [], []
    
    for elem in mixed_lst:
        match elem:
            case int():
                integers.append(elem)
            case float():
                floats.append(elem)
            case str():
                strings.append(elem)
                
    return floats, integers, strings

def mixed_lists_separation2(mixed_lst):
    floats = []
    integers = []
    strings = []
    
    for elem in mixed_lst:
        if isinstance(elem,int):
            integers.append(elem)
        elif isinstance(elem,float):
            floats.append(elem)    
        elif isinstance(elem,str):
            strings.append(elem)
    return floats, integers, strings    
    
aa,bb,cc =  mixed_lists_separation2(["mixed_lst",1,2.3,5,-9,"Hello","there", " ",4,44,-3.2])
a,b,c =  mixed_lists_separation(["mixed_lst",1,2.3,5,-9,"Hello","there", " ",4,44,-3.2])
print(aa," , ", bb, " , ", cc)
print()
print(a)
print(b)
print(c)
```
