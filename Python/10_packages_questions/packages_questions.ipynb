```python
# Import the `math` module and use it to calculate the square root of 25 and the sine of 90 degrees.

import math

from sympy import degree

def calc_square_root(num):
    return math.sqrt(num)

def calc_sine(degree):
    return math.sin(math.radians(degree))

print(calc_square_root(25))
print(calc_sine(90))
```

```python
# Import the `datetime` module with an alias and use it to print the current date and time.

import datetime as dt

print(dt.datetime.now())

```

```python
# Import the `randint` function from the `random` module and use it to generate a random integer between 1 and 100.


from random import randint

print(randint(1, 100))
```

```python
# Import the `sqrt` and `pow` functions from the `math` module
# and use them to calculate the square root of 16 and 2 raised to the power of 3.

from math import sqrt, pow

print(sqrt(16))
print(pow(2, 3))

```

```python
# Use the `os` module to create a new directory,
# list the contents of the current directory,
# and remove the newly created directory.

import os

os.mkdir('new_directory')
print(os.listdir('.'))
os.rmdir('new_directory')
```

```python
# Use the `sys` module to print the Python version currently in use
# and the command-line arguments passed to the script.

import sys

print(f"Python version: {sys.version}")
print(f"Command-line arguments: {sys.argv}")
```

```python
# Use the `math` module to calculate the greatest common divisor (GCD) of two numbers
# and the factorial of a number.

from math import gcd, factorial

print(gcd(48, 18))
print(factorial(5))
```

```python
# Use the `datetime` module to print the current date,
# calculate the date 100 days from today,
# and determine the day of the week for a given date.

import datetime

today = datetime.date.today()
date_100_days_later = today + datetime.timedelta(days=100)
given_date = datetime.date(2023, 10, 1)

print(f"Today's date: {today}")
print(f"Date 100 days from today: {date_100_days_later}")  
print(f"Day of the week for the given date: {given_date.strftime('%A')}")
```

```python
# Use the `random` module to generate a list of 5 random numbers between 1 and 50
# and shuffle the elements of a list.

import random

random_numbers = [random.randint(1, 50) for _ in range(5)]
print(f"Random numbers: {random_numbers}")
```

```python
# Write code to import and use the functions from `mypackage`
# without explicitly importing `module1` and `module2`.

import mypackage

print(mypackage.module2.multiply(3, 5))
```

```python
# Another way, more specific to avoid importing the whole package:
from mypackage import module1, module2

print(module1.add(3, 5))
print(module2.multiply(3, 5))
```

```python
# and finally - we can also:

from mypackage import *
print(module1.add(3, 5))
```

```python
from mypackage.subpackage import module3

print(module3.multi_multi(3, 5))
```

```python
# Write code that attempts to import a non-existent function from `mypackage`
# and gracefully handles the import error by printing an error message.

from mypackage import module1

try:
    module1.copycopy(2)
except AttributeError as e:
    print(f"Error: {e}")
```

```python
try:
    from mypackage import multi_multi
    print(multi_multi(2, 3))
except ImportError as e:
    print(f"Error importing function: {e}")
```
