```python
# Create a dictionary with the first 10 positive integers as keys and their squares as values. Print the dictionary.

dict = {key:key**2 for key in range(1,11)}
print(dict)
```

```python
# Print the value of the key 5 and the keys of the dictionary created in Assignment 1.

print(dict[5])
```

```python
# Add a new key-value pair (11, 121) to the dictionary created in Assignment 1 and then remove the key-value pair with key 1. Print the modified dictionary.
if dict.get(11) == None:
    dict[11] = 11**2
else:
    print("11 is already in dict")    
print(dict)

if dict.get(1) != None:
    del dict[1]
else:
    print("dict[1] was already deleted")  
print(dict)
```

```python
# Iterate over the dictionary created in Assignment 1 and print each key-value pair.

for key, val in dict.items():
    print(f"key: {key}, val: {val}")
```

```python
# Create a new dictionary containing the cubes of the first 10 positive integers using a dictionary comprehension. Print the new dictionary.

dict_cubes = {key: key**3 for key in range(1,11)}
print(dict_cubes)
```

```python
# Create two dictionaries: one with keys as the first 5 positive integers and values as their squares, and another with keys as the next 5 positive integers and values as their squares. Merge these dictionaries into a single dictionary and print it.

dict_first_5 = {key:key**2 for key in range(1,6)}
dict_last_5 = {key:key**2 for key in range(6,11)}

print("dict_first_5: ")
print(dict_first_5)
print("dict_last_5: ")
print(dict_last_5)

dict_merge = dict_first_5 | dict_last_5
print(dict_merge)
```

```python
def print_dict(dict):
    for key, val in dict.items():
        print(f"{key} : {val}")
```

```python
# Create a dictionary where the keys are the first 5 positive integers
# and the values are lists containing the first 5 multiples of the key. Print the dictionary.


multiples = {}

for i in range(1,6):
    multiples[i] = [i * j for j in range(1,6)]
    
# 1 : [1*1, 1*2, 1*3, 1*4, 1*5]
# 2 : [2*1, 2*2, 2*3, 2*4, 2*5]
# 3 : [3*1, 3*2, 3*3, 3*4, 3*5]
# 4 : [4*1, 4*2, 4*3, 4*4, 4*5]
# 5 : [5*1, 5*2, 5*3, 5*4, 5*5]

# print(multiples)

print_dict(multiples)
```

```python
# Create a dictionary where the keys are the first 5 positive integers
# and the values are tuples containing the key and its square. Print the dictionary.

dict_tpl = {}

for i in range(1,6):
    dict_tpl[i] = (i, i**2)

print_dict(dict_tpl)
```

```python
# Create a dictionary with the first 5 positive integers as keys and their squares as values.
# Convert the dictionary to a list of tuples and print it.

print("dict_to_lst:")
dict_to_lst = {key:key**2 for key in range(1,6)}
print_dict(dict_to_lst)

print()

lst_from_dict = [(key,val) for (key,val) in dict_to_lst.items()]
print("lst_from_dict:")
print(lst_from_dict)

print()

lst_from_dict_2 = [(items) for items in dict_to_lst.items()]
print("lst_from_dict_2:")
print(lst_from_dict_2)
```

```python
# Create a dictionary with the first 10 positive integers as keys and their squares as values.
# Create a new dictionary containing only the key-value pairs where the key is even.
# Print the new dictionary.

dict_even_odd = {key:key**2 for key in range(1,11)}

dict_only_even = {key:val for key in dict_even_odd if key % 2 == 0}

print_dict(dict_only_even)

```

```python
# Create a dictionary with the first 5 positive integers as keys and their squares as values.
# Create a new dictionary with keys and values swapped. Print the new dictionary.

dict = {key:key**2 for key in range(1,11)}
dict_swap = {val:key for key,val in dict.items()}

print_dict(dict_swap)
```

```python
# Create a default dictionary where each key has a default value of an empty list.
# Add some elements to the lists and print the dictionary.

default_dict = {key:[None] for key in range(1,6)}
print_dict(default_dict)

print()

default_dict[3].append("Hello")
default_dict[1].append(2)
default_dict[6] = True
print_dict(default_dict)
```

```python
# Write a function that takes a string and returns a dictionary with the count of each character in the string.
# Print the dictionary.

def str_to_count_chars_dict(word):
    count_chars = {key: word.count(key) for key in word}
    return count_chars

word = "Hello BF! I Love you!"
returned_dict = str_to_count_chars_dict(word)
print_dict(returned_dict)
```

```python
# Create a dictionary representing a book with keys 'title', 'author', 'year', and 'genre'.
# Convert the dictionary to a JSON string and print it.

import json

book = {
    "Title":None,
    "Author":None,
    "Year" :None,
    "Genre" :None
}

print_dict(book)

json_str = json.dumps(book)
print(json_str)
```
