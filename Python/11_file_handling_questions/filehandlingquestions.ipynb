```python
# Write a function that reads the contents of a file named `sample.txt` and prints each line.

def read_file2(file_name):
    try:
        with open(file_name, 'r') as file:
            for line in file:
                print(line.strip()) # strip() is used to remove any leading/trailing whitespace, including newlines
    except FileNotFoundError:
        print("The file 'sample.txt' was not found.")
    except Exception as e:
        print(f"An error occurred: {e}")


def read_file(file_name):
    with open(file_name, 'r') as file:
        for line in file:
            print(line, end='')  # end='' prevents adding an extra newline since lines already contain newlines

read_file2('sample.txt')
read_file('sample.txt')
```

```python
# Write a function that writes a list of strings to a file named `output.txt`, with each string on a new line.

# 'w' starts every time from the beginning of the file and overwrites the content
def write_to_file2(file_name, lst):
    try:
        with open(file_name, 'w') as file:
            for item in lst:
                file.write(str(item) + ' ')  # Add a newline character after each string
    except Exception as e:
        print(f"An error occurred while writing to the file: {e}")


def write_to_file(file_name, lst):
    with open(file_name, 'w') as file: 
        for item in lst:
            file.write(str(item) + " ")

file_name = "output.txt"
lst = ["so 1","last", "time", "I need to be", "The", 1, "who takes you home", "\n"]
lst2 = ["Hello", "World", "This", "is", "a", "test"]

write_to_file(file_name, lst)
read_file(file_name)

write_to_file2(file_name, lst2)
read_file(file_name)
```

```python
# Write a function that copies the contents of a file named `source.txt` to a new file named `destination.txt`.

import shutil

# First place efficiency
def file_copy(from_file1, to_file2):
    shutil.copy(from_file1, to_file2)

# Second place
def file_copy2(from_file1, to_file2):
    with (
    open(from_file1, 'r') as file1,
    open(to_file2, 'w') as file2
    ):
        for line in file1: # a line is the default "item" in python when iterating over a file object
            file2.write(line)

# Third place - NOT RECOMMENDED
# This method reads the entire content of the file into memory
def file_copy3(from_file1, to_file2):
    with open(from_file1, 'r') as file1:
        content = file1.read()  # read the entire content of the file
    with open(to_file2, 'w') as file2:
        file2.write(content)  # write the entire content to the new file
         

source_file1 = "sample.txt"
source_file2 = "output.txt"
dest_file = "dest.txt"
file_copy(source_file1, dest_file)
read_file(dest_file)   

file_copy2(source_file2, dest_file)            
read_file(dest_file)  
```

```python
# Write a function that appends a given string to the end of a file named `log.txt`.

def append_to_file(file_name, text):
    try:
        with open(file_name, 'a') as file:  
            file.write(text + '\n')  
    except Exception as e:
        print(f"An error occurred while appending to the file: {e}")

def append_to_eof(a_str, file_name):
    with open(file_name, 'a') as file:
        file.write(a_str)

a_str = "OLA!!!!"    
a_str2 = "This is a new line"  
file_name = "log.txt"

append_to_file(file_name, a_str)
read_file(file_name)
append_to_eof(a_str2, file_name)
read_file(file_name)
```

```python
# Write a function that reads the contents of a file named `sample.txt`
# and returns the number of words in the file.

def count_words_in_file(file_name):
    try:
        with open(file_name, 'r') as file:
            content = file.read()  # Read the entire content of the file
            words = content.split()  # Split the content into words based on whitespace
            return len(words)  # Return the number of words
    except FileNotFoundError:
        print(f"The file '{file_name}' was not found.")
        return None
    except Exception as e:
        print(f"An error occurred: {e}")
        return None
    
def num_of_words_in(file_name):
    sum = 0
    with open(file_name, 'r') as file:
        for line in file:
           words_list = line.split()  # Split the line into words based on whitespace
           sum += len(words_list)
    return sum

document_file = "sample.txt"
print(num_of_words_in(document_file))
```

```python
# Write a function that finds and replaces all occurrences
# of a given word in a file named `sample.txt` with another word.

import os

def replace_all_occurences(file_name, find_str, replace_str):
    with (
        open(file_name, 'r') as file_find,
        open("temp_file.txt", 'w') as file_temp
        ):
            for line in file_find:
                file_temp.write(line.replace(find_str, replace_str))
                
    # change between file_find and file_temp and delete file_temp
    os.replace("temp_file.txt", file_name)  # Rename the temporary file to the original file name


old_word = "OLA"
new_word = "ALOHA"
replace_all_occurences("sample.txt", old_word, new_word)  
read_file("sample.txt")
```

```python
# Write a function that reads the contents of a file named `reverse.txt`
# and prints each line in reverse order.

def print_file_lines_backwards(file_name):
    with open(file_name, 'r') as file:
        for line in file:
            print(line.strip()[::-1])

file_name = "sample.txt"
print_file_lines_backwards(file_name)
```

```python
# Write a function that reads the contents of a file named `stats.txt` 
# and returns the number of lines, words, and characters in the file.

def num_of_lines_words_chars_in(file_name):
    lines, words, chars = 0, 0, 0
    
    with open(file_name, 'r') as file:
        for line in file:
            lines += 1 
            words +=len(line.split())
            chars += len(line)
    return lines, words, chars

file_name = "sample.txt"
print(num_of_lines_words_chars_in(file_name))
```

```python
# Write a function that merges the contents of multiple files into a single file named `merged.txt`.

def merge_files_into_one(list_of_files, output_file = "merged.txt"): # if name is not given
    with open(output_file, 'w') as out_file:
        for a_file in list_of_files:
            with open(a_file, 'r') as file:
                for line in file:
                    out_file.write(line)
            out_file.write("\n")        

file_name = "all.txt"
list_of_files = ["sample.txt", "output.txt", "log.txt", "dest.txt"]
merge_files_into_one(list_of_files, file_name)
read_file(file_name)
```
