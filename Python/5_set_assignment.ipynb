```python
# Create a set with the first 10 positive integers. Print the set.

set_positive = set(range(1,11))
print(set_positive)

# Add the number 11 to the set created in Assignment 1. Then remove the number 1 from the set. Print the modified set.

set_positive.add(11)
set_positive.discard(1)

print(set_positive)
```

```python
# Create two sets: one with the first 5 positive integers and another with the first 5 even integers
# Perform and print the results of union, intersection, difference, and symmetric difference operations on these sets.

set_5_positive = set(range(1,6))
set_5_even = set(range(2,11,2))

print(f"set_5_positive: {set_5_positive}")
print(f"set_5_even: {set_5_even}")

set_union = set_5_even | set_5_positive
# set_union = set_5_even.union(set_5_positive) # that is the same

set_intersection = set_5_even & set_5_positive
# set_intersection = set_5_even.intersection(set_5_positive) # that is the same

set_even_diff_positive = set_5_even - set_5_positive
# set_even_diff_positive = set_5_even.difference(set_5_positive)  # that is the same

set_positive_diff_even = set_5_positive - set_5_even
# set_positive_diff_even = set_5_positive.difference(set_5_even)  # that is the same

set_symetric_diff = set_5_even ^ set_5_positive
# set_symetric_diff = set_5_even.symmetric_difference(set_5_positive)  # that is the same

print(f"set_union: {set_union}")
print(f"set_intersection: {set_intersection}")
print(f"set_even_diff_positive: {set_even_diff_positive}")
print(f"set_positive_diff_even: {set_positive_diff_even}")
print(f"set_symetric_diff: {set_symetric_diff}")

```

```python
# Create a new set containing the squares of the first 10 positive integers using a set comprehension. Print the new set.

set_square = {x**2 for x in range(1,11)}
print(set_square)
```

```python
# Create a new set containing only the even numbers from the set created in Assignment 1 using a set comprehension. Print the new set.

even_set = {x for x in set_positive if x%2 == 0}
print(even_set)
```

```python
# Create a set with duplicate elements and remove the duplicates using set methods. Print the modified set.

lst = [1,2,2,4,4,4]
set_from_lst = set(lst)
print(set_from_lst)
```

```python
# Create two sets: one with the first 5 positive integers and another with the first 3 positive integers.
# Check if the second set is a subset of the first set and if the first set is a superset of the second set. Print the results.

a_set = set(range(1,6))
b_set = set(range(1,4))

print(f"a_set : {a_set}")
print(f"b_set : {b_set}")

print(f"Is a_set a subset of b_set? {a_set.issubset(b_set)}")
print(f"Is a_set a superset of b_set? {a_set.issuperset(b_set)}")

print(f"Is b_set a subset of a_set? {b_set.issubset(a_set)}")
print(f"Is b_set a superset of a_set? {b_set.issuperset(a_set)}")

```

```python
# Create a frozenset with the first 5 positive integers. Print the frozenset.


# A frozenset is an immutable version of a set, meaning its elements cannot be changed after creation.
# but they do not guarantee element positions.

frozen_set = frozenset(range(1,6))
print(frozen_set) # this odd print is just how it is (and is used in order to the difference between a regular set and a frozen one)

# a normal mutable set
normal_set = {1,4,6,7,54} 
print(normal_set) # a normal set print

```

```python
# Create a set with the first 5 positive integers.
# Convert it to a list, append the number 6, and convert it back to a set. Print the resulting set.

five_positive_int_set = set(range(1,6))
five_positive_int_list = list(five_positive_int_set)
five_positive_int_list.append(6)
five_positive_int_set = set(five_positive_int_list)

print(five_positive_int_list)
print(five_positive_int_set)


# Extra:
five_positive_int_list.append(8)
five_positive_int_set.add(9)

print(five_positive_int_list)
print(five_positive_int_set)
```

```python
# Create a set and remove elements from it until it is empty. Print the set after each removal.

set_bb = {44,6,3,-1,4}

while len(set_bb) > 0:
    removed_val = set_bb.pop()
    print(f"set: {set_bb}, \t removed_val : {removed_val}")
```

```python
# Create two sets and update the first set with the symmetric difference of the two sets. Print the modified first set.

set_one = {1,2,3,4,5}
set_two= {4,5,6,7,8}

set_one = set_one ^ set_two

print(set_one)
```

```python
# Create a set and test if certain elements are present in the set. Print the results.

set_cc = {33,5,2,1,34}
nums = (1,2,3,4)

result = {(val,val in nums) for val in set_cc}
print(result)
print(type(result))
```
