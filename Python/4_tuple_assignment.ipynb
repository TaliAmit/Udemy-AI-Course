```python
# Create a tuple with the first 10 positive integers
ten_positive = tuple(range(1,11))
print("ten_positive:", ten_positive)

# Print the first, middle, and last elements of the tuple created in Assignment 1.
first, *middle, last = ten_positive

print("First element:", first)
print("Middle elements:", middle)
print("Last element:", last)

# Print the first three elements, the last three elements, and the elements from index 2 to 5 of the tuple created in Assignment 1.

first_three_elem = ten_positive[0:3]
last_three_elem = ten_positive[-3:]
some_between_elem = ten_positive[2:6]

print("First three elements:", first_three_elem)
print("Last three elements:", last_three_elem)
print("Some elements in between:", some_between_elem)
```

```python
# Create a nested tuple representing a 3x3 matrix and print the matrix. Access and print the element at the second row and third column.

# Tuples are immutable (cannot be changed after creation)
# so I can't use a for loop to "append" items

matrix = ((1, 2, 3),
          (4, 5, 6),    
          (7, 8, 9))

print("Matrix:", matrix)

# import numpy as np
# print(np.array(matrix)) 

# Access and print the element at the second row and third column
element = matrix[1][2]
print("Element at 2nd row and 3rd column:", element)
```

```python
# Concatenate two tuples: (1, 2, 3) and (4, 5, 6). Print the resulting tuple.

# Tuples are immutable, meaning you cannot modify their contents (or replace existing elements) after the tuple is created.

first_tuple = (1,2,3)
second_tuple = (4,5,6)

concutenated_tuple = first_tuple + second_tuple
print("Concatenated tuple:", concutenated_tuple)
```

```python
# Create a tuple with duplicate elements and count the occurrences of an element. Find the index of the first occurrence of an element in the tuple.

import random

duplicates_tuple = tuple(random.choices(range(1,6), k=7))
print("duplicates_tuple:", duplicates_tuple)

for i, val in enumerate(duplicates_tuple):
    count = duplicates_tuple.count(val)
    if count > 1:
        print(f"{val} appears {count} times.\nIts first occurrence is at index {i}")
        break
```

```python
# Create a tuple containing 3 tuples, each with 3 elements. Print the tuple of tuples.

import random

rows = 3
cols = 3

nested_tuple = tuple( tuple( random.choices(range(1,11), k = cols) ) for _ in range(rows) )
print(nested_tuple)
```

```python
# Create a tuple with the characters of a string. Join the tuple elements into a single string. Print the string.

tpl_str = tuple("Hello you")
print(tpl_str)

str = "".join(tpl_str)
print(str)
```

```python
# Create a nested tuple and iterate over the elements, printing each element.

import random

rows = 2
cols = 3

nested_tpl = tuple( tuple( random.choices(range(1,11), k = cols)) for _ in range(rows) )
print(f"{nested_tpl} \n")

for row_elem in range(rows):
    print(nested_tpl[row_elem])

print()

for i in range(rows):
    for j in range(cols):
        print(nested_tpl[i][j], end=", ")
    print()    

```

```python
# Write functions that take a tuple and return the minimum, maximum, and sum of the elements. Print the results for a sample tuple.


def min_max_sum_from_tuple(tpl):
    total = sum(tpl)
    min_val = min(tpl)
    max_val = max(tpl)

    return total, min_val, max_val

print(min_max_sum_from_tuple((1,2,3,4)))
```
