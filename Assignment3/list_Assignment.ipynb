```python
# Create a list of random numbers and sort it in ascending and descending order.
# Remove the duplicates from the list and print the modified list.

import random

lst2 = random.choices(range(1,10), k=20)            # with duplicates
print(f"lst2: {lst2}");

lst2_asending = sorted(lst2)                        # asending order
print(f"lst2_asending: {lst2_asending}");

# lst2_desending - Best way:
lst2_desending = lst2_asending[ : : -1]             # descending order
print(f"lst2_desending: {lst2_desending}");

# Other ways:
# 1st way:
    # lst2_desending = lst2_asending.copy()
    # lst2_desending.reverse()   

    # this won't work : lst2_desending = lst2_asending.copy().reverse()
    # because reverse() is inplace and returns None, so it will assign None to lst2_desending
# 2nd way:                           
    # lst2_desending = sorted(lst2, reverse = True)
    # this one is sorting all over again, so it's less efficient than the first way

# lst2_without_duplicates - Best way:
lst2_without_duplicates = list(set(lst2))          # without duplicates
print(f"lst2_without_duplicates: {lst2_without_duplicates}")
```

```python
# Write a function that takes a 3x3 matrix (nested list) as input and returns its transpose.
# Print the original and transposed matrices.

import random
import numpy as np

cols = 3
rows = 3

mtrx = [random.choices(range(1,10), k=cols) for _ in range(rows)]
print(f"mtrx:\n{mtrx}")
print(type(mtrx), np.shape(mtrx))

mtrx = np.array(mtrx)  # not necessary in this exercise because np.transpose() can also work on regular nested list
print(f"mtrx:\n{mtrx}")
print(type(mtrx), np.shape(mtrx))

transposed_mtrx = np.transpose(mtrx)

print(f"transposed_mtrx: \n{transposed_mtrx}") 
print(type(transposed_mtrx), np.shape(transposed_mtrx))
```

```python
# Write a function that takes a nested list and flattens it into a single list.
# Print the original and flattened lists.

import numpy as np

rows = 2
cols = 3

nested_list = [ random.choices(range(1,10), k=cols) for _ in range(rows) ]
print(f"Original nested list: {nested_list}")

flattened_list = np.array(nested_list).flatten().tolist()  # this is the best way to flatten a nested list, it's very efficient and easy to read
print(f"Flattened list: {flattened_list}")

# other way:
def flatten_list(lst):
    flattened2 = []
    for sublist in lst:
        flattened2.extend(sublist)
    lst = flattened2    


nested_list = [ random.choices(range(1,10), k=cols) for _ in range(rows) ]
flatten_list(nested_list)
```

```python
# Create a list of the first 10 positive integers
# Remove the elements at indices 2, 4, and 6
# insert the element '99' at index 5. Print the modified list.

lst = list(range(1,11))
del lst[2:7:2] # 2,4,6
print(lst)

lst1 = lst.copy()
lst1[5] = 99    # overwrite the element at index 5 (9) with 99
print(f"lst1: {lst1}")

lst.insert(5, 99)
# add a new element
# it will be inserted before the element at index 5 (9)
# the element at index 5 (9) will be moved to index 6
print(f"lst: {lst}")
```

```python
# Write a function that takes a list and returns a new list with the elements in reverse order. Print the original and reversed lists.
def reversed_list(lst):
    return lst[ : : -1]

lst1 = list(range(1,16))[::3]
print(f"Original list: {lst1}")
print(f"Reversed list: {reversed_list(lst1)}")
```

```python
# Write a function that rotates a list by n positions. Print the original and rotated lists.

import collections

def rotate_right_k(lst, k):
    lst1_left = lst[:-k]	# [1,2,3,4,5] 
    lst1_right = lst[-k:]	# [6,7] 
    
    lst[:] = lst1_right + lst1_left

lst1=[1,2,3,4,5,6,7]
k=2

# solution using collections.deque:
rotate_right_k(lst1, k)
print(lst1)

def rotate_list(lst, n):
    d = collections.deque(lst)
    d.rotate(n)
    return list(d)

lst = list(range(1,11))[::-2]
print(f"Original list: {lst}")

print(f"Rotated list: {rotate_list(lst, 2)}")
```

```python
# Write a function that takes two lists
# return a new list containing only the elements that are present in both lists.
# Print the intersected list.

lst1 = [1, 2, 3, 4, 5]
lst2 = [4, 5, 6, 7, 8]

def intersection(lst1, lst2):
    # 1st solution: works, not clean and not efficient 
    # intersect_lst = [val for val in lst1 for val2 in lst2 if val == val2]

    # 2nd solution: works, clean but not efficient
    #intersect_lst = [val for val in lst1 if val in lst2] 

    # 3rd solution: clean and efficient
    # set is using a hash map to find elements quickly - o(1)
    set_lst2 = set(lst2)
    intersect_lst = [ val for val in lst1 if val in set_lst2 ]

    return intersect_lst

print(intersection(lst1, lst2))
```
